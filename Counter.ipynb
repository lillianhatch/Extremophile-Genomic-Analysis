```python
import pandas as pd

#loading the thermobase csv
thermoBase = pd.read_csv('/content/ThermoBase_ver_1.0_2022.csv', encoding='latin1')
print(thermoBase.columns)
print(thermoBase.head())
```

```python
#Filtering the dataset so only bacteria are included

thermoBacteria = thermoBase[thermoBase['Domain'] == 'Bacteria']

print(thermoBacteria.head())
```

```python
#Grouping the bacteria by family and counting species within each family. Print results with largest to smallest families

familyCount = thermoBacteria.groupby('Family').size().sort_values(ascending=False)
print(familyCount)
```

```python
#Saving the list of Bacillaceae family species within a .txt for later use

bacillaceaeData = thermoBacteria[thermoBacteria['Family'] == 'Bacillaceae']
listBacillaceae = bacillaceaeData['Name']
listBacillaceae = listBacillaceae.dropna().str.strip()
listBacillaceae.to_csv('/content/drive/MyDrive/Capstone/Data/listBacillaceae.txt', index=False, header=False, sep='\n')

print("Successfully exported " + str(len(listBacillaceae)) + " species to csv")
```
