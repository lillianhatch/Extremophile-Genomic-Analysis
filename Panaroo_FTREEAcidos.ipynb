```python
#panaroo processing in which the .gff files are read in
##genome alignment is output as well as gene presence matrix
!mkdir -p /content/local_gffs
!cp /content/drive/MyDrive/Capstone/Data/Acidos/prokka_output/*/*.gff /content/local_gffs/

!conda run -n panaroo panaroo \
    -i /content/local_gffs/*.gff \
    -o /content/panaroo_local_output \
    --clean-mode moderate \
    -a core \
    --aligner mafft \
    -t 24 \
    --remove-invalid-genes

!mkdir -p /content/drive/MyDrive/Capstone/Data/Acidos/panaroo_output_v3
!cp -r /content/panaroo_local_output/* /content/drive/MyDrive/Capstone/Data/Acidos/panaroo_output_v3/
```

```python
!mkdir -p /content/local_gffs
!cp /content/drive/MyDrive/Capstone/Data/Acidos/prokka_output/*/*.gff /content/local_gffs/

!conda run -n panaroo panaroo \
    -i /content/local_gffs/*.gff \
    -o /content/panaroo_local_output \
    --clean-mode sensitive \
    --refind-mode off \
    -a core \
    --aligner mafft \
    -t 24 \
    --remove-invalid-genes

!mkdir -p /content/drive/MyDrive/Capstone/Data/Acidos/panaroo_output_v4
!cp -r /content/panaroo_local_output/* /content/drive/MyDrive/Capstone/Data/Acidos/panaroo_output_v4/
```
