```python
fastaSource = "/content/drive/MyDrive/Capstone/Data/Thermos/FASTAS/data"
outputFiles = "/content/drive/MyDrive/Capstone/Data/Thermos/prokka_output"
os.makedirs(outputFiles, exist_ok=True)

accessionFolders = sorted(os.listdir(fastaSource))
print(f"Found {len(accessionFolders)} accession folders")

for accession in accessionFolders:
    sampleOutdir = os.path.join(outputFiles, accession)
    gffCheck = os.path.join(sampleOutdir, f"{accession}.gff")

    # Skip if already done (resume capability)
    if os.path.exists(gffCheck):
        print(f"Skipping {accession} — already completed")
        continue

    sequenceFiles = glob.glob(os.path.join(fastaSource, accession, f"{accession}_*_genomic.fna"))
    if not sequenceFiles:
        print(f"WARNING: no genome FASTA found for {accession}, skipping")
        continue

    sequenceFiles = sequenceFiles[0]
    print(f"Running Prokka on {accession}...")
    !prokka --outdir {sampleOutdir} --prefix {accession} --cpus 0 --fast --force {sequenceFiles}

print("Batch complete.")
```
